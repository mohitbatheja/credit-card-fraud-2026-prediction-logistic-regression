```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
x = df.drop(["transaction_id","is_fraud"],axis=1)
y = df["is_fraud"]
```

```python
categorical_cols = [
    "merchant_category",
    "card_type",
    "auth_method",
    "channel",
    "device_type",
    "is_foreign_transaction",
    "is_new_merchant",
    "used_vpn",
    "ip_country_mismatch",
    "billing_shipping_mismatch",
    "is_ai_generated_scam_attempt"
]
x = pd.get_dummies(
    x,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)
```

```python
x
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,y,test_size = 0.20, stratify=y, random_state=42
)
```

```python
model = LogisticRegression( 
    max_iter=1000,
    class_weight="balanced")
```

```python
model
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
comparions = pd.DataFrame({
    "Actual": y_test,
    "Prediction": y_pred
})
```

```python
comparions
```

```python
accuracy = accuracy_score(y_test,y_pred)*100
precision = precision_score(y_test,y_pred)
recall = recall_score(y_test,y_pred)*100
f1_score = f1_score(y_test,y_pred)*100
```

```python
print(f"accuracy: {accuracy:.2f}%")
print("precision:",precision)
print("recall:",recall)
print("f1_score:",f1_score)
```

```python
cm = confusion_matrix(y_test,y_pred)
```

```python
cm
```

```python
print(classification_report(y_test,y_pred))
```

The Logistic Regression model with balanced class weights achieved 85.42% accuracy and 88.24% recall for fraud detection. It successfully identified 60 out of 68 fraudulent transactions, although the low precision of 9.45% indicates a relatively high false-positive rate.
